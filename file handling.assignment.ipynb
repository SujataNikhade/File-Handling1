```python
#Q8.Remove Blank Lines Create a script to remove all blank lines from a file.
f=open("fil.txt","w")
f.write("""
Hello this is first line.

This file has some blank lines.


Let's clean them up.
""")
f.close()
f=open("fil.txt","r")
data=f.readlines()
f.close()
f=open("fil.txt","w")
for i in data:
    if i.strip()!="":
        f.write(i)
        print(i,end="")
f.close()
```

```python
#Q9.Read a File in Reverse Order Print the lines of a file in reverse order (last line first).
f=open("sample.txt","w")
f.write("\nhello,\nsujju\n how\n are\n you!\n")
f.close()
f=open("sample.txt","r")
data=f.readlines()
i=len(data)-1
while i>=0:
    print(data[i],end=" ")
    i-=1
f.close()
```

```python
#Q10.Line Frequency Counter Count the frequency of each line in a file (useful for analyzing logs or data files).
f=open("sample.txt","w")
f.write("hello how are you sujju you are looking good today sujju hello sujju")
f.close()
f=open("sample.txt","r")
data=f.read().split()
s={}
for i in data:
    line=i.strip()
    if line in s:
        s[line]+=1
    else:
        s[line]=1
for i,count in s.items():
    print(i,count)
f.close()
```

```python
#Q2. You are asked to design a file system that allows you to create new paths and associate them with different values.
file_system=[]
def create_path(path,value):
    if path=="/" or not path:
        return False
    for p,v in file_system:
        if p==path:
            return False
    idx=path.rfind("/")
    pattern=path[:idx]
    if pattern!="" and not any(p==pattern for p,v in file_system):
        return False
    file_system.append([path,value])
    return True
def get(path):
    for p,v in file_system:
        if p==path:
            return v
    return -1
print(create_path("/a",5))
print(get("/a"))
```

```python
# we have file of 10 email we need to mask email to paste another file
import re
def masked(mail):
    pattern="([a-zA-Z])([a-zA-Z0-9._]+)(@)([a-zA-Z]+)(\.)([a-zA-Z]+)"
    match=re.search(pattern,mail)
    if match:
        g1=match.group(1)
        g2=len(match.group(2))*"#"
        g3=match.group(3)
        g4=match.group(4)
        g5=match.group(5)
        g6=match.group(6)
        mask=g1+g2+g3+g4+g5+g6
        f2.write(mask+"\n")
f2=open("maskedmail.txt","w")
f3=open("raw_email","r")
data=f3.read()
new_data=data.split("\n")
for i in new_data:
    masked(i)
f3.close()
f2.close()
f2=open("maskedmail.txt","r")
data=f2.read()
print(data)
```
